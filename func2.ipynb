```python
def is_prime(n: int) -> bool:
    """Check whether an integer is a prime number.

    A prime number is an integer greater than 1 that has exactly two positive
    divisors: 1 and itself.

    Args:
        n (int): The integer to check for primality.

    Returns:
        bool: True if n is a prime number, otherwise False.

    Raises:
        ValueError: If n is not an integer.

    Example:
        >>> is_prime(2)
        True
        >>> is_prime(10)
        False
        >>> is_prime(17)
        True
    """
    if not isinstance(n, int):
        raise ValueError("n must be an integer.")

    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False

    divisor = 3
    while divisor * divisor <= n:
        if n % divisor == 0:
            return False
        divisor += 2

    return True


def sum_of_primes(numbers: list[int]) -> int:
    """Calculate the sum of all prime numbers in a list.

    The function validates that the input is a list and that all elements are
    integers before summing only the prime values.

    Args:
        numbers (list[int]): A list of integers to evaluate.

    Returns:
        int: The sum of all prime numbers found in numbers.

    Raises:
        TypeError: If numbers is not a list.
        ValueError: If any element in numbers is not an integer.

    Example:
        >>> sum_of_primes([1, 2, 3, 4, 5, 10])
        10
        >>> sum_of_primes([4, 6, 8])
        0
    """
    if not isinstance(numbers, list):
        raise TypeError("numbers must be a list.")

    total = 0
    for number in numbers:
        if not isinstance(number, int):
            raise ValueError("All elements in numbers must be integers.")
        if is_prime(number):
            total += number

    return total
```
